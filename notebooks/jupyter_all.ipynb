```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from xgboost import XGBClassifier

from notebooks.helper.commoncode import create_data, get_metrics
import pandas as pd
import os,joblib
```

```python
df = pd.read_csv('../data/train.csv')
X_train, X_test, y_train, y_test, scaler, feature_columns = create_data(df)
model = LogisticRegression(
    max_iter=5000,          # allows convergence
    # multi_class='auto',     # handles multiclass automatically
    solver='lbfgs',         # stable solver for multiclass
    random_state=42
)

model.fit(X_train, y_train)
y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test)
print(get_metrics(y_test, y_pred, y_prob))
```

```python
model = KNeighborsClassifier(
n_neighbors=5,      # important: better than default 5 for this dataset
metric="minkowski",
p=2                 # Euclidean distance
)

model.fit(X_train, y_train)

y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test)
print(get_metrics(y_test, y_pred, y_prob))
```

```python
"""Decision Tree"""

model = DecisionTreeClassifier(
        criterion="gini",     # default and works well for this dataset
        max_depth=None,       # allow tree to grow fully
        random_state=42
    )


model.fit(X_train, y_train)
y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test)
print(get_metrics(y_test, y_pred, y_prob))
```

```python
"""Naive Bayes"""
model = GaussianNB()

model.fit(X_train, y_train)
y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test)
print(get_metrics(y_test, y_pred, y_prob))
```

```python
"""Random Forest"""

model = RandomForestClassifier(
    n_estimators=200,   # stabilizes results
    random_state=42,
    n_jobs=-1           # uses all CPU cores (faster training)
)

model.fit(X_train, y_train)
y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test)
print(get_metrics(y_test, y_pred, y_prob))
```

```python
"""XGBoost"""

model = XGBClassifier(
    objective='multi:softprob',
    num_class=4,
    eval_metric='mlogloss',
    # use_label_encoder=False,
    n_estimators=200,
    learning_rate=0.1,
    max_depth=6,
    random_state=42
)

model.fit(X_train, y_train)
y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test)
print(get_metrics(y_test, y_pred, y_prob))
```
